**Findings:**

- Only `Type_of_Loan` has real nulls (1,426). Other missing values are written as text placeholders (`_` in `Credit_Mix`; one placeholder SSN shared by 703 customers), so `isna()` does not find them
- Names repeat across up to 6 customers and carry stray letters. Identifiers (`Customer_ID`, `Name`, `SSN`) will not be features.


## 3. Candidate keys

```python
def check_key(df, key):
    """
    Check df against a primary key or composite key by checking 
    for total row count, missing values, and duplicates.
    """
    print("rows:", len(df))
    
    # Check the key columns for missing values
    print("blank key rows:", df[key].isna().any(axis=1).sum())
    
    # Count how many rows share the exact same key values as an earlier row
    print("duplicate key rows:", df.duplicated(subset=key).sum())

for name, df in [("attributes", attributes), ("financials", financials), ("clickstream", clickstream)]:
    print(f"--- {name}")
    check_key(df, ["Customer_ID", "snapshot_date"])
    print("distinct customers:", df["Customer_ID"].nunique())
```

**Findings:** 
- No blank or duplicate keys
- Attributes and financials hold one row per customer (12,500 each), not monthly snapshots
- Clickstream holds 24 monthly rows for each of 8,974 customers
- LMS holds 12,500 loans, one per customer

## 4. Dates

```python
def check_dates(values):
    """
    Check date strings range, uniqueness and format
    """
    
    # Attempt to convert the series to datetime objects using the specific format. Convert to NaT if error.
    parsed = pd.to_datetime(values, format="%Y-%m-%d", errors="coerce")
    
    # Print statistics
    print("min:", values.min(), " max:", values.max(), " distinct:", values.nunique())
    
    # Count missing/invalid data:
    # 1. Missing values (NaN, None)
    # 2. Not a valid YYYY-MM-DD date
    print("blank:", values.isna().sum(), " not a valid YYYY-MM-DD date:", (parsed.isna() & values.notna()).sum())

for name, df in [("attributes", attributes), ("financials", financials), ("clickstream", clickstream), ("lms", lms)]:
    print(f"--- {name} snapshot_date")
    check_dates(df["snapshot_date"])

print("--- lms loan_start_date")
check_dates(lms["loan_start_date"])
```

**Findings:** 
- All dates are valid month-starts, so every source is monthly

- Attributes' monthly counts match LMS's new loans per month (e.g. 530 in 2023-01), suggesting each customer's attributes/financials are taken in their loan application month (TBC)
- Clickstream covers the same 8,974 customers every month, including months after their loan started
